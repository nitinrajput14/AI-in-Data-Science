1. Armstrong Number in a Range.  .
Print all Armstrong numbers between 1 and N.

```python
#Armstrong number

print("Armstrong numbers between 1 and 500:")
num = 0
while num <=500:
  num+=1
  str_num = str(num)
  n = len(str_num)

  arm_num = sum(int(digit)**n for digit in str_num)
  if arm_num == num:
    print(num,arm_num)
```

2. Happy Number \
A number is called Happy if repeatedly replacing it with the sum of the squares of its digits eventually reaches 1.

```python
num = int(input("Enter a numnber to start happy number loop: "))
ori_num = num

while num != 1 and num!= 4:
  num = sum(int(digit)**2 for digit in str(num))
  if num == 1:
    break

if num == 1 :
  print(f"{ori_num} : is a Happy Number")
else:
  print(f"{ori_num} : is a not Happy Number")
```

4. Password Strength Checker (Using Loops) \
Check whether a password contains:\
 Uppercase\
 Lowercase\
 Digit\
 Special Character\
 Minimum 8 characters

```python
password = input("Enter a password: ")

has_upper = False
has_lower = False
has_digit = False
has_special = False

has_valid_length = len(password) >= 8

for char in password:
  if char.isupper():
    has_upper = True
  elif char.islower():
    has_lower = True
  elif char.isdigit():
    has_digit = True
  elif not char.isalnum():
    has_special = True

if  has_valid_length and has_upper  and has_lower and has_digit  and has_special  :
  print(f"{password} : is a Strong password")
else:
  print("Weak password : Missing requirements")
  if not has_valid_length : print("- Must be at least 8 characters long")
  if not has_upper :         print("- Missing Uppercase")
  if not has_lower  :        print("- Missing Lowercase")
  if not has_digit   :       print("- Missing digit")
  if not has_special  :      print("- Missing special char.")


```

6. Print Pascal's Triangle\
 Print the first N rows of Pascal's Triangle.\
 Example (N = 5):

```python
n = 5
for i in range(n):
    val = 1
    for j in range(i + 1):
        print(val, end=" ")
        # Condition to calculate the next number in the same row
        val = val * (i - j) // (j + 1)
    print()  # New line for the next row
```
